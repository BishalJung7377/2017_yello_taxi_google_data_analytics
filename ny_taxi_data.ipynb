###  6 Core Practices of EDA:
#### 1. Discovering:

        Get familiar with the data (column names, data types, volume).

        Ask initial questions (What’s here? What might be missing?).

#### 2. Structuring:

        Organize and categorize the raw data.

        Prepare it for easier visualization and analysis.

        Be aware of bias—grouping data inaccurately can skew results.

#### 3. Cleaning:

        Fix errors like:

        Missing values

        Duplicates

        Typos

        Outliers


#### 4. Joining:

        Combine your dataset with other sources to add context or fill gaps.

        Example: Add demographic data or missing categories.

#### 5.  Validating:

        Check for:

        Consistent formatting

        Accuracy of cleaning

        High data quality

        Use tools like Python, R, etc., to automate this process.

#### 6.  Presenting:

        Share findings with stakeholders via:

        Cleaned datasets

        Charts, graphs, dashboards

        Use accessible communication (alt text, captions) and ethical visualization (no misleading charts).

```python
### lets convert tpep_pickup_datetime and tpep_dropoff_datetime intoo date time object

df['tpep_pickup_datetime'] = pd.to_datetime(df['tpep_pickup_datetime'])

df['tpep_dropoff_datetime'] = pd.to_datetime(df['tpep_dropoff_datetime'])
```

```python
df.nunique()
```

```python
## categoriging the pickup times 
def categorize_time_of_day(start_time):
    hour = start_time.hour
    if 5 <= hour <= 12:
        return 'Morning'
    elif 12<= hour <=17:
        return  'Afternoon'
    elif 17<= hour <= 21:
        return 'Evening'
    else:
        return "Night"
```

```python
df['time_of_day'] = df['tpep_pickup_datetime'].apply(categorize_time_of_day)
```

### extracting months from the pickup date


```python
df['month'] = df['tpep_pickup_datetime'].dt.strftime('%B')
```

```python
### lets create a new features called peak_hours
peak_hours = [8,9,10,17,18,19] ## based on historical data 8,9,10 am usually considered peak hours
df['hour'] = df['tpep_pickup_datetime'].dt.hour
df['peak_hours'] = df['hour'].apply(lambda x: 'peak' if x in peak_hours else "off-peak")
```

```python
### lets create a new features called weekoftheday
df['week_of_day'] = df['tpep_pickup_datetime'].dt.strftime('%a')
```

```python
## lets try to bin the distance features 
def distance_class(x):
    if x<2:
        return 'Very Short Distance'
    elif x<5:
        return 'Short Distance'
    elif x<10:
        return 'Medium Distance'
    else:
        return "Long Distance"
```

```python
df['distance_category'] = df['trip_distance'].apply(distance_class)
```
